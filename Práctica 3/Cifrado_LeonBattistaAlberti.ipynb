```python
import string


ALFABETO = string.ascii_uppercase + string.digits  # A-Z y 0-9
TAMANO_DISCO = len(ALFABETO) # Será 36

def cifrar(texto, desplazamiento_inicial=0, periodo=4):
    resultado = []
    giro = desplazamiento_inicial
    contador = 0
    for letra in texto.upper():
        if letra not in ALFABETO:
            resultado.append(letra)  # espacios/signos no se cifran
            continue
        idx = ALFABETO.index(letra)
        idx_cifrado = (idx + giro) % TAMANO_DISCO
        resultado.append(ALFABETO[idx_cifrado])
        contador += 1
        if contador % periodo == 0:
            giro += 1  # se gira el disco interior una posición

    return "".join(resultado)

def descifrar(texto_cifrado, desplazamiento_inicial=0, periodo=4):
    resultado = []
    giro = desplazamiento_inicial
    contador = 0
    for letra in texto_cifrado.upper():
        if letra not in ALFABETO:
            resultado.append(letra)
            continue
        idx = ALFABETO.index(letra)
        idx_claro = (idx - giro) % TAMANO_DISCO
        resultado.append(ALFABETO[idx_claro])
        contador += 1
        if contador % periodo == 0:
            giro += 1
    return "".join(resultado)

texto = input("Texto: ")
cifrado = cifrar(texto, 1, 4)
descifrado = descifrar(cifrado, 1, 4)
print("\nCifrado (Ck):", cifrado)
print("Descifrado (Dk):", descifrado)
```
