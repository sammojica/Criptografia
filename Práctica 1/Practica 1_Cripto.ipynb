```python
A = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"

texto = input("Texto: ").upper()
k = int(input("Llave k: "))

cifrado = ""
for char in texto:
    if char in A:
        cifrado += A[(A.index(char) + k) % 26]
    else:
        cifrado += char

descifrado = ""
for char in cifrado:
    if char in A:
        descifrado += A[(A.index(char) - k) % 26]
    else:
        descifrado += char

print("\nCifrado (Ck):", cifrado)
print("Descifrado (Dk):", descifrado)
```
