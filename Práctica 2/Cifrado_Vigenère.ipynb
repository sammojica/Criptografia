```python
A = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"

texto = input("Texto: ").upper()
clave = input("Llave k (palabra): ").upper()

# --- CIFRADO ---

cifrado = ""
idx_clave = 0

for char in texto:
    if char in A:
        # k_i es el valor numérico de la letra actual de la clave
        k_i = A.index(clave[idx_clave % len(clave)])
        cifrado += A[(A.index(char) + k_i) % 26]
        idx_clave += 1  # Avanza solo si se cifró un carácter válido
    else:
        cifrado += char

# --- DESCIFRADO ---
descifrado = ""
idx_clave = 0

for char in cifrado:
    if char in A:
        k_i = A.index(clave[idx_clave % len(clave)])
        descifrado += A[(A.index(char) - k_i) % 26]
        idx_clave += 1
    else:
        descifrado += char

print("\nCifrado (Ck):", cifrado)
print("Descifrado (Dk):", descifrado)
```
